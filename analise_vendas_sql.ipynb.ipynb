```python
import pandas as pd
import sqlite3

# Conectar ao banco de dados (cria automaticamente)
conexao = sqlite3.connect('banco_vendas.db')
print("✅ Banco conectado com sucesso!")
```

```python
# Dados de exemplo (pode substituir pelo arquivo da empresa)
dados = pd.DataFrame({
    "Produto": ["Notebook", "Celular", "Tablet", "Fone", "Carregador"],
    "Categoria": ["Informática", "Telefonia", "Informática", "Áudio", "Acessórios"],
    "Quantidade": [5, 12, 8, 25, 30],
    "Valor_Unitario": [3500, 2500, 1800, 150, 80]
})

# Salvar como tabela no banco
dados.to_sql("vendas", conexao, index=False, if_exists="replace")
print("✅ Tabela 'vendas' criada com", len(dados), "registros")
dados
```

```python
# 1️⃣ Ver todos os registros
consulta = """
SELECT *
FROM vendas
"""
todos = pd.read_sql(consulta, conexao)
print("=== 1. Todos os registros ===")
todos
```

```python
# 2️⃣ Produtos acima de R$ 1.000
consulta = """
SELECT
    Produto AS "Nome do Produto",
    Categoria,
    Valor_Unitario AS "Preço"
FROM vendas
WHERE Valor_Unitario > 1000
ORDER BY Valor_Unitario DESC
"""
produtos_caros = pd.read_sql(consulta, conexao)
print("=== 2. Produtos acima de R$ 1.000 ===")
produtos_caros
```

```python
# 3️⃣ Resumo por categoria
consulta = """
SELECT
    Categoria,
    COUNT(*) AS "Quantidade de Produtos",
    SUM(Quantidade) AS "Total em Estoque",
    AVG(Valor_Unitario) AS "Valor Médio"
FROM vendas
GROUP BY Categoria
ORDER BY "Valor Médio" DESC
"""
por_categoria = pd.read_sql(consulta, conexao)
print("=== 3. Desempenho por Categoria ===")
por_categoria
```

```python
# 4️⃣ Top 3 produtos com maior valor total
consulta = """
SELECT
    Produto,
    Categoria,
    Quantidade,
    Valor_Unitario,
    Quantidade * Valor_Unitario AS "Valor Total em Estoque"
FROM vendas
ORDER BY "Valor Total em Estoque" DESC
LIMIT 3
"""
top3 = pd.read_sql(consulta, conexao)
print("=== 4. Top 3 Mais Valiosos ===")
top3
```

```python
# 5️⃣ Relatório completo para entrega
consulta = """
SELECT
    Produto,
    Categoria,
    Quantidade,
    Valor_Unitario,
    Quantidade * Valor_Unitario AS "Valor Total"
FROM vendas
ORDER BY "Valor Total" DESC
"""
relatorio_final = pd.read_sql(consulta, conexao)

# Exibir resumo
total_geral = relatorio_final["Valor Total"].sum()
media = relatorio_final["Valor Total"].mean()

print("=" * 50)
print("📊 RELATÓRIO FINAL DE VENDAS")
print("=" * 50)
print(f"Total de produtos: {len(relatorio_final)}")
print(f"Valor geral do estoque: R$ {total_geral:,.2f}")
print(f"Média por produto: R$ {media:,.2f}")
print("=" * 50)

relatorio_final
```

```python
# Salvar arquivo pronto
relatorio_final.to_csv("relatorio_completo.csv", index=False)
print("✅ Arquivo 'relatorio_completo.csv' GERADO!")
print("\n🎉 Projeto concluído com sucesso! Pronto para entrega.")
```
